```python
def even_odd(test_value):
    if test_value % 2 == 0:
        return "That is an even number."
    else: return "That is an odd number."


test_value = int(input("Type in your number"))

print(even_odd(test_value))
    
```

## Exercise 3
##### Password Strength Checker

```python
password = input("Create a strong password here")
has_digit = False
spec_char = "!£$%^&*"
has_special = False
    
if len(password) < 8:
    print ("Password is too short")
for character in password:
    if character.isdigit():
        has_digit = True
if has_digit == False:
    print ("Password must include at lease one number")

for character in password:
    if character in spec_char:
        has_special = True
if has_special == False: 
    print("Password must contain a special character such as !£$%^&*")

if len(password) >= 8 and has_digit and has_special:
    print ("You have created a strong password")
```

## Exercise 5
##### Temperature Conversion Table

```python
def temp_conversion():
    start = int(input("What is your starting temperature in Celsius?"))
    end = int(input("What is your ending temperature in Celsius?"))

    print (f"{"Celsius":<10}{"Farenheit":<10}")
    
    for i in range(start, end+1):
        Farenheit = (i * 9/5)+32
        print (f"{i:<10}{Farenheit:<10}")

temp_conversion()
```

## Exercise 6
##### Factorial Function with Input Check

```python
def factorial():
    number = int(input("Type in your number"))
    if number < 0:
            print ("Number cannot be negative")
    else:
        total = 1
        for i in range(1, number+1):
            total *=i
        print (f"The factorial of {number} is {total}.")
            
factorial()  
```

## Exercise 7
##### Prime Number Checker

```python
def prime_checker():
    number = int(input("Type in your number"))
    if number <= 0:
        print("Number cannot be zero or less")
    else:
        for i in range (2, number):
            if number % i == 0:
                print("This is not a prime number")
                break
        else:
            print("This is a prime number")

prime_checker()
```
