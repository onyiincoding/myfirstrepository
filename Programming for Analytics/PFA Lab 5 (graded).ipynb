#### Question 1

```python
student_grades = []
try: 
    with open("grades.txt", "r") as file:
        reader = file.read().split("\n")
    
        for i in reader:
            if i:
                split_reader = i.split(",")
    
                grade_dict = {"name": split_reader[0], "grade": split_reader[1]}
    
                student_grades.append(grade_dict)

except OSError:
    print("File error encountered")
```

#### Question 2

```python
## creating a list of the first names
import requests

first_names = []
for i in student_grades:
    firstname = i["name"].split(" ")[0]

    first_names.append(firstname)

    
```

```python
## creating an age key in the dictionaries
for i in student_grades:
    i["age"] = ""

## looping through the first names as parameters
try: 
    for i in first_names:
        agifyurl = "https://api.agify.io"
        agify_params = {"name": i}
    
        agify_response = requests.get(agifyurl, agify_params)
        predicted_age = agify_response.json()
    
        ## inserting the age from Agify into the dictionaries
        i["age"] = predicted_age["age"]
except KeyError:
    print("Key could not be accessed")
```

#### Question 3

```python
## caclulating the average
total = 0
for i in student_grades:
    total += int(i["grade"])

    average = total/len(student_grades)

print(f"The average grade is {average}.")
```

```python
## inserting status based on average into the dictionaries
for i in student_grades:
    if int(i["grade"]) >= average:
        i["status"] = "Over"
    else:
        i["status"]= "Under"

print(student_grades)
```

#### Question 4

```python
## creating a json file containing enriched information of students
try:
    import json
    
    with open("students_enriched.json", "w") as json_file:
        json.dump(student_grades, json_file)

except OSError:
    print("File error encountered")
```

#### Question 5

```python
## making a nested lists containing the dictionaries converted into lists

over_students = []
for i in student_grades: 
    over_students.append(list(i.values()))

print(over_students)
```

```python
## creating a csv file containing only the students with the 'Over' status
import csv

try:
    with open("Over_students.csv", "w") as csvfile:
        csvwriter = csv.writer(csvfile)
        csvwriter.writerow (["name", "grade", "status", "age"])
    
        for i in over_students:
            if i[2] == "Over":
                csvwriter.writerow(i)
except OSError as e:
    print("File error encountered")


```
