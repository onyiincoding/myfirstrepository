### inputting results and printing them out

```python
print ("Hello! Let's talk about your rectangle, shall we?")
your_length = int(input("what is the length in cm?"))
your_breadth = int(input("what is the breadth in cm?"))

your_area = your_length * your_breadth
your_pmeter = 2 * (your_length + your_breadth)

print (f"Great! The area is {your_area}cm\u00B2 and the perimeter is {your_pmeter}cm.")
```

# Extension Challenge

```python
print ("Hello! Welcome to the BMI Calculator. Let's get started!")

your_weight = int(input("What is your weight in kg?"))
your_height = float(input("What is your height in meters?"))

your_bmi = your_weight / your_height**2

print (f"Perfect! Your BMI is {your_bmi:.2f}.")
```
