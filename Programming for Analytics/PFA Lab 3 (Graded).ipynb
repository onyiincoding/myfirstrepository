#### Create a list of 3 Student Dictionaries
##### containing name, age and grade

```python
students = [{"Name": "Andra Bay", "Age": int(20), "Grade": int(70)},\
            {"Name": "Charles Dam", "Age": int(21), "Grade": int(80)},\
            {"Name": "Ezra Finn", "Age": int(22), "Grade": int(90)}]
```

```python
students
```

#### Print the average grade

```python
total = 0
for i in students: 
    total += i["Grade"]

average_grade = total/len(students)

print(average_grade)
```

#### Print the names of students who scored above the average

```python
for i in students:
    if i["Grade"] > average_grade:
        print (f"{i["Name"]} scored above the average grade")
```

#### Convert names to uppercase and print them

```python
for i in students:
    i["Name"] = i["Name"].upper()
```

```python
students
```

#### Checking for grade in list

```python
student_name = input("Please type in your full name")

try: 
    for i in students:
        if student_name.upper() == i["Name"]:
            print (i["Grade"])
            break
    else: print ("Record not found")
except ValueError: 
    print ("Invalid input")
```
